## **Data Loading**

```python
import pandas as pd
df = pd.read_csv("iris.csv")
df.head(5)
```

## **Data Visualization**

```python
import matplotlib.pyplot as plt
x = df['species']
y = df['sepal_length']

plt.plot(x, y, marker='o', linestyle='-', label='Data Points')
plt.grid(True)
plt.xlabel("Species")
plt.ylabel("Sepal_length")
plt.title("Iris")
plt.legend()
plt.show
```
